```python
from pathlib import Path

import numpy as np
import pandas as pd
```

```python
data_dir = Path("./scripts_survey_wifi/data")
csv_dir = data_dir / "archive/csv"

active = lambda author: csv_dir / f"active/active_L3_{author}.csv"
calibration = lambda author: csv_dir / f"calibration/kalibrasi_L3_{author}.csv"
passive = lambda author: csv_dir / f"passive/passive_L3_{author}.csv"

deva_active_path = active("Deva")
dinda_active_path = active("Dinda")
merged_active_path = data_dir / "active_L3.csv"

maul_cal_path = calibration("Maulana")
taqi_cal_path = calibration("Taqiya")
merged_cal_path = data_dir / "calibration_L3.csv"

maul_pass_path = passive("Maulana")
taqi_pass_path = passive("Taqiya")
merged_pass_path = data_dir / "passive_L3.csv"
```

# Active

```python
deva_active = pd.read_csv(deva_active_path, sep=",", skip_blank_lines=True)
dinda_active = pd.read_csv(dinda_active_path, sep=",", skip_blank_lines=True)
merged_active = pd.concat([dinda_active, deva_active], ignore_index=True)
merged_active.head()
```

```python
merged_active.info()
```

```python
merged_active["timestamp"] = pd.to_datetime(merged_active["timestamp"])
merged_active.info()
```

```python
# sort merged berdasarkan timestamp
tmp = merged_active.sort_values(by="timestamp", ascending=True).reset_index().drop(columns="index")
tmp.head(20)
```

```python
merged_active = tmp.copy()
merged_active.head(10)
```

```python
merged_active.to_csv(merged_active_path, sep=",", index=False)
pd.read_csv(merged_active_path).head(10)
```

# Passive

```python
taqi_pass = pd.read_csv(taqi_pass_path, sep=",")
maul_pass = pd.read_csv(maul_pass_path, sep=",")

merged_pass = pd.concat([taqi_pass, maul_pass], ignore_index=True)
merged_pass.head()
```

```python
merged_pass.info()
```

```python
merged_pass["timestamp"] = pd.to_datetime(merged_pass["timestamp"])

merged_pass.info()
```

```python
merged_pass["lantai"] = 3
merged_pass.head()
```

```python
merged_pass.to_csv(merged_pass_path, sep=",", index=False)

pd.read_csv(merged_pass_path).head(10)
```
